## Grounded SAM 2 (Grounding DINO + SAM 2.1)

```python
%pip install -q transformers accelerate sam2 Pillow tqdm
```

```python
# Imports & config
import json, random
from pathlib import Path
import numpy as np
import torch
from PIL import Image, ImageDraw
from tqdm import tqdm
from transformers import AutoProcessor, AutoModelForZeroShotObjectDetection
from sam2.sam2_image_predictor import SAM2ImagePredictor
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

GDINO_ID       = "IDEA-Research/grounding-dino-base"
SAM2_ID        = "facebook/sam2.1-hiera-large"
IMAGE_DIR      = "/content/facades_dataset_v0"
OUTPUT_DIR     = "/content/gsam2_output"
DEVICE         = "cuda" if torch.cuda.is_available() else "cpu"

TEXT_PROMPT    = "window . balcony . door ."
BOX_THRESHOLD  = 0.40
TEXT_THRESHOLD = 0.35
SAM2_BOX_BATCH = 8

CLASS_COLORS   = {"window": (0, 100, 255), "door": (220, 50, 50), "balcony": (50, 200, 50)}
DEFAULT_COLOR  = (180, 180, 0)
```

```python

# Load models
gdino_processor = AutoProcessor.from_pretrained(GDINO_ID)
gdino_model     = AutoModelForZeroShotObjectDetection.from_pretrained(GDINO_ID).to(DEVICE)
sam2_predictor  = SAM2ImagePredictor.from_pretrained(SAM2_ID)
print(f"Models loaded on {DEVICE}")

# Helpers
def detect(image: Image.Image):
    inputs = gdino_processor(
        images=image,
        text=[[c.strip() for c in TEXT_PROMPT.replace(".", "").split()]],
        return_tensors="pt",
    ).to(DEVICE)
    with torch.no_grad():
        outputs = gdino_model(**inputs)
    results = gdino_processor.post_process_grounded_object_detection(
        outputs, inputs.input_ids,
        threshold=BOX_THRESHOLD,
        text_threshold=TEXT_THRESHOLD,
        target_sizes=[image.size[::-1]],
    )[0]
    return results["boxes"].cpu().numpy(), results["scores"].cpu().numpy(), results["labels"]


def segment(image: Image.Image, boxes: np.ndarray) -> np.ndarray:
    """SAM 2.1 с батчингом боксов — предотвращает OOM на изображениях с большим числом объектов"""
    if len(boxes) == 0:
        H, W = image.size[1], image.size[0]
        return np.zeros((0, H, W), dtype=bool)

    img_arr = np.array(image.convert("RGB"))
    all_masks = []

    with torch.inference_mode(), torch.autocast(DEVICE, dtype=torch.bfloat16):
        sam2_predictor.set_image(img_arr)
        for i in range(0, len(boxes), SAM2_BOX_BATCH):
            batch = boxes[i : i + SAM2_BOX_BATCH]
            masks, _, _ = sam2_predictor.predict(
                point_coords=None,
                point_labels=None,
                box=batch,
                multimask_output=False,
            )
            if masks.ndim == 4:
                masks = masks[:, 0]
            all_masks.append(masks.astype(bool))
            torch.cuda.empty_cache()

    return np.concatenate(all_masks, axis=0)


def color_for(label: str):
    for k, v in CLASS_COLORS.items():
        if k in label.lower():
            return v
    return DEFAULT_COLOR


def draw_overlay(image: Image.Image, masks, labels, scores) -> Image.Image:
    canvas = np.array(image.convert("RGB"), dtype=np.float32)
    for mask, label in zip(masks, labels):
        color = np.array(color_for(label), dtype=np.float32)
        canvas[mask] = canvas[mask] * 0.45 + color * 0.55
    result = Image.fromarray(canvas.clip(0, 255).astype(np.uint8))
    draw = ImageDraw.Draw(result)
    for mask, label, score in zip(masks, labels, scores):
        ys, xs = np.where(mask)
        if len(xs) == 0:
            continue
        x0, y0, x1, y1 = xs.min(), ys.min(), xs.max(), ys.max()
        c = color_for(label)
        draw.rectangle([x0, y0, x1, y1], outline=c, width=2)
        draw.text((x0 + 2, y0 + 2), f"{label} {score:.2f}", fill=c)
    return result


def mask_to_rle(mask: np.ndarray) -> dict:
    flat = mask.flatten(order="F").astype(np.uint8)
    counts, prev, run = [], 0, 0
    for v in flat:
        if v == prev:
            run += 1
        else:
            counts.append(run); run = 1; prev = v
    counts.append(run)
    return {"counts": counts, "size": list(mask.shape)}
```

```python
import torchvision
import math

def dynamic_detect(image: Image.Image, target_size=1000, overlap_ratio=0.2):
    W, H = image.size

    if W <= target_size and H <= target_size:
        return detect(image)

    stride_w = int(target_size * (1 - overlap_ratio))
    stride_h = int(target_size * (1 - overlap_ratio))

    x_starts = list(range(0, W, stride_w))
    y_starts = list(range(0, H, stride_h))

    all_boxes, all_scores, all_labels = [], [], []

    for y in y_starts:
        for x in x_starts:
            x_end = min(x + target_size, W)
            y_end = min(y + target_size, H)

            x_start = max(0, x_end - target_size)
            y_start = max(0, y_end - target_size)

            crop = image.crop((x_start, y_start, x_end, y_end))

            inputs = gdino_processor(
                images=crop,
                text=[[c.strip() for c in TEXT_PROMPT.replace(".", "").split()]],
                return_tensors="pt",
            ).to(DEVICE)

            with torch.no_grad():
                outputs = gdino_model(**inputs)

            results = gdino_processor.post_process_grounded_object_detection(
                outputs, inputs.input_ids,
                threshold=BOX_THRESHOLD,
                text_threshold=TEXT_THRESHOLD,
                target_sizes=[crop.size[::-1]],
            )[0]

            boxes = results["boxes"].cpu().numpy()
            scores = results["scores"].cpu().numpy()
            labels = results["labels"]

            if len(boxes) > 0:
                boxes[:, 0] += x_start
                boxes[:, 1] += y_start
                boxes[:, 2] += x_start
                boxes[:, 3] += y_start

                all_boxes.append(boxes)
                all_scores.append(scores)
                all_labels.extend(labels)

    if not all_boxes:
        return np.empty((0, 4)), np.array([]), []

    boxes_np = np.vstack(all_boxes)
    scores_np = np.concatenate(all_scores)

    boxes_tensor = torch.tensor(boxes_np, dtype=torch.float32)
    scores_tensor = torch.tensor(scores_np, dtype=torch.float32)

    keep_indices = torchvision.ops.nms(boxes_tensor, scores_tensor, iou_threshold=0.4)
    keep_indices = keep_indices.numpy()

    final_boxes = boxes_np[keep_indices]
    final_scores = scores_np[keep_indices]
    final_labels = [all_labels[i] for i in keep_indices]

    return final_boxes, final_scores, final_labels
```

```python
# Visual test on 5 random images
for img_path in sample:
    image = Image.open(img_path).convert("RGB")
    boxes, scores, labels = dynamic_detect(image)
    masks   = segment(image, boxes)
    overlay = draw_overlay(image, masks, labels, scores)

    fig, axes = plt.subplots(1, 2, figsize=(16, 7))
    fig.suptitle(f"{img_path.name}  |  {len(boxes)} detections", fontsize=11)
    axes[0].imshow(image);   axes[0].set_title("Original"); axes[0].axis("off")
    axes[1].imshow(overlay); axes[1].set_title("Detections + Masks"); axes[1].axis("off")

    patches = [mpatches.Patch(color=np.array(v)/255., label=k) for k, v in CLASS_COLORS.items()]
    fig.legend(handles=patches, loc="lower center", ncol=3, fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
# Full dataset inference
overlay_dir = Path(OUTPUT_DIR) / "overlays"
mask_dir    = Path(OUTPUT_DIR) / "masks"
overlay_dir.mkdir(parents=True, exist_ok=True)
mask_dir.mkdir(parents=True, exist_ok=True)

coco_images, coco_annotations, ann_id = [], [], 1

for img_id, img_path in enumerate(tqdm(paths, desc="Inference"), start=1):
    try:
        image = Image.open(img_path).convert("RGB")
        W, H  = image.size
        boxes, scores, labels = detect(image)
        masks = segment(image, boxes)

        draw_overlay(image, masks, labels, scores).save(
            overlay_dir / f"{img_path.stem}_overlay.png")

        for i, (mask, label, score) in enumerate(zip(masks, labels, scores)):
            Image.fromarray(mask.astype(np.uint8) * 255, mode="L").save(
                mask_dir / f"{img_path.stem}_{i:03d}_{label}.png")

        coco_images.append({"id": img_id, "file_name": img_path.name, "width": W, "height": H})
        for mask, label, score in zip(masks, labels, scores):
            ys, xs = np.where(mask)
            if len(xs) == 0:
                continue
            x0, y0, x1, y1 = int(xs.min()), int(ys.min()), int(xs.max()), int(ys.max())
            coco_annotations.append({
                "id": ann_id, "image_id": img_id,
                "category_name": label,
                "bbox":   [x0, y0, x1 - x0, y1 - y0],
                "score":  float(score),
                "segmentation": mask_to_rle(mask),
                "area":   int(mask.sum()),
                "iscrowd": 0,
            })
            ann_id += 1

        torch.cuda.empty_cache()

    except Exception as e:
        print(f"ERROR {img_path.name}: {e}")
        torch.cuda.empty_cache()

coco_out = {
    "images": coco_images,
    "annotations": coco_annotations,
    "categories": [{"id": i+1, "name": k} for i, k in enumerate(CLASS_COLORS)],
}
(Path(OUTPUT_DIR) / "results.json").write_text(json.dumps(coco_out, indent=2))

print(f"\nDone. {len(paths)} images | {ann_id-1} instances")
print(f"  overlays/    — RGB с масками и боксами")
print(f"  masks/       — бинарные маски per instance")
print(f"  results.json — COCO JSON для импорта в CVAT")
```

## SegFormer-B5, fine-tuned on CMP Facades




```python
import os, random
import numpy as np
import torch
import torch.nn.functional as F
from pathlib import Path
from PIL import Image
from tqdm.auto import tqdm
from transformers import SegformerImageProcessor, SegformerForSemanticSegmentation
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

MODEL_ID   = "Xpitfire/segformer-finetuned-segments-cmp-facade"
IMAGE_DIR  = "/content/facades_dataset_v0"
OUTPUT_DIR = "/content/segformer_output"
DEVICE     = "cuda" if torch.cuda.is_available() else "cpu"

LABELS = {
    0: "background", 1: "facade",  2: "molding", 3: "cornice",
    4: "pillar",     5: "window",  6: "door",    7: "sill",
    8: "blind",      9: "balcony", 10: "shop",   11: "deco",
    12: "unknown",
}
PALETTE = np.array([
    [0,0,0], [128,128,128], [139,90,43],  [255,165,0],
    [0,128,0], [0,0,255],  [255,0,0],    [255,255,0],
    [0,255,255], [255,0,255], [128,0,128], [0,255,0],
    [64,64,64],
], dtype=np.uint8)
```

```python
processor = SegformerImageProcessor.from_pretrained(MODEL_ID)
model = SegformerForSemanticSegmentation.from_pretrained(MODEL_ID).eval().to(DEVICE)
print(f"Loaded on {DEVICE} | classes: {model.config.num_labels}")
```

```python
def predict(image: Image.Image) -> np.ndarray:
    inputs = processor(images=image, return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        logits = model(**inputs).logits
    upsampled = F.interpolate(logits, size=(image.size[1], image.size[0]),
                              mode="bilinear", align_corners=False)
    return upsampled.argmax(dim=1).squeeze(0).cpu().numpy().astype(np.int64)

def to_rgb(seg: np.ndarray) -> np.ndarray:
    seg_clipped = np.clip(seg, 0, len(PALETTE) - 1)
    return PALETTE[seg_clipped]

def coverage(seg: np.ndarray) -> dict:
    total = seg.size
    ids, counts = np.unique(seg, return_counts=True)
    return {LABELS.get(int(i), f"cls_{i}"): round(c / total * 100, 2)
            for i, c in zip(ids, counts)}

def blend(image: Image.Image, seg: np.ndarray, alpha: float = 0.5) -> np.ndarray:
    return (alpha * np.array(image) + (1 - alpha) * to_rgb(seg)).clip(0,255).astype(np.uint8)
```

```python
paths = sorted(Path(IMAGE_DIR).glob("*.png"))
sample = random.sample(paths, 5)

for img_path in sample:
    image = Image.open(img_path).convert("RGB")
    seg   = predict(image)
    cov   = coverage(seg)

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle(img_path.name, fontsize=11)
    axes[0].imshow(image);              axes[0].set_title("Original");  axes[0].axis("off")
    axes[1].imshow(to_rgb(seg));        axes[1].set_title("Mask");       axes[1].axis("off")
    axes[2].imshow(blend(image, seg));  axes[2].set_title("Overlay");    axes[2].axis("off")

    patches = [mpatches.Patch(color=PALETTE[min(i, len(PALETTE)-1)]/255.,
                               label=f"{LABELS.get(int(i), f'cls_{i}')} {cov.get(LABELS.get(int(i), f'cls_{i}'), 0):.1f}%")
               for i in np.unique(seg)]
    fig.legend(handles=patches, loc="lower center", ncol=6, fontsize=8, bbox_to_anchor=(0.5, -0.04))
    plt.tight_layout()
    plt.show()

    print(f"{img_path.name}  |  "
          + "  ".join(f"{c}: {cov.get(c, 0):.1f}%" for c in ["window","door","balcony"]))
```

```python
mask_dir    = Path(OUTPUT_DIR) / "masks"
overlay_dir = Path(OUTPUT_DIR) / "overlays"
mask_dir.mkdir(parents=True, exist_ok=True)
overlay_dir.mkdir(parents=True, exist_ok=True)

csv_rows = ["filename," + ",".join(LABELS.values())]

for img_path in tqdm(paths, desc="Inference"):
    try:
        image = Image.open(img_path).convert("RGB")
        seg   = predict(image)

        Image.fromarray(seg.astype(np.uint8), mode="L").save(
            mask_dir / f"{img_path.stem}_mask.png")

        Image.fromarray(blend(image, seg)).save(
            overlay_dir / f"{img_path.stem}_overlay.png")

        total = seg.size
        row = [img_path.name] + [str(round(np.sum(seg == i) / total * 100, 4))
                                  for i in range(len(LABELS))]
        csv_rows.append(",".join(row))

    except Exception as e:
        print(f"ERROR {img_path.name}: {e}")

(Path(OUTPUT_DIR) / "coverage.csv").write_text("\n".join(csv_rows))
print(f"\nDone. {len(paths)} images → {OUTPUT_DIR}")
print(f"  masks/     — grayscale, pixel = class index [0-12]")
print(f"  overlays/  — RGB blend")
print(f"  coverage.csv")
```
